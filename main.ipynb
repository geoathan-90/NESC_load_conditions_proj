```python
import xarray as xr

ds2 = xr.open_dataset("data_1.nc")

print(ds2)

for var in ds2.data_vars:
    print(f"\n--- {var} ---")
    print(ds2[var].attrs)
```

```python
ds2 = ds2.assign_coords(
    longitude=((ds2.longitude + 180) % 360) - 180
).sortby("longitude")
```

```python
print(ds2.longitude.min().item())
print(ds2.longitude.max().item())
```

```python
print(
    "Latitude match:",
    np.allclose(ds.latitude, ds2.latitude)
)

print(
    "Longitude match:",
    np.allclose(ds.longitude, ds2.longitude)
)

print(
    "Time match:",
    np.array_equal(ds.valid_time, ds2.valid_time)
)
```

```python
# Convert ds2 from 0–360 longitude convention
ds2 = ds2.assign_coords(
    longitude=((ds2.longitude + 180) % 360) - 180
).sortby("longitude")

# Verify that the physical grids coincide
assert np.allclose(ds.latitude.values, ds2.latitude.values)
assert np.allclose(ds.longitude.values, ds2.longitude.values)
assert np.array_equal(ds.valid_time.values, ds2.valid_time.values)

# Remove tiny floating-point discrepancies by adopting ds's
# coordinate labels as the canonical grid
ds2 = ds2.assign_coords(
    latitude=ds.latitude,
    longitude=ds.longitude,
    valid_time=ds.valid_time,
)

# Now require exact alignment
weather = xr.merge(
    [ds, ds2],
    join="exact",
    compat="no_conflicts",
)

assert weather.sizes["longitude"] == 71
assert weather.sizes["latitude"] == 51
assert weather.sizes["valid_time"] == 744
```

```python
print("snowc:", weather.snowc.min().item(),
                  weather.snowc.max().item())

print("sde:", weather.sde.min().item(),
                weather.sde.max().item())
```

```python
weather["t2m_c"] = weather["t2m"] - 273.15
weather["t2m_c"].attrs["units"] = "°C"
weather["t2m_c"].attrs["long_name"] = "2 metre temperature"

weather["wind10"] = np.hypot(
    weather["u10"],
    weather["v10"],
)
weather["wind10"].attrs["units"] = "m/s"
weather["wind10"].attrs["long_name"] = "10 metre wind speed"

weather["sde"] = weather["sde"].clip(min=0)
weather["snowc"] = weather["snowc"].clip(min=0, max=100)
```

```python
weather = weather.rio.set_spatial_dims(
    x_dim="longitude",
    y_dim="latitude"
)

weather = weather.rio.write_crs("EPSG:4326")
```

```python
weather.to_netcdf(
    "weather_2020_01_montana.nc"
)
```
