```python
import xarray as xr
import numpy as np
import pandas as pd
import rioxarray

print("xarray:", xr.__version__)
```

```python
ds = xr.open_dataset("data_0.nc")

ds
```

```python
for var in ds.data_vars:
    print(f"\n--- {var} ---")
    print(ds[var].attrs)
```

```python
import numpy as np

# Temperature: K -> °C
t2m_c = ds["t2m"] - 273.15
t2m_c.attrs["units"] = "°C"
t2m_c.attrs["long_name"] = "2 metre temperature"

# 10 m wind speed
wind10 = np.hypot(ds["u10"], ds["v10"])
wind10.attrs["units"] = "m/s"
wind10.attrs["long_name"] = "10 metre wind speed"
```

```python
print(f"T minimum: {t2m_c.min().item():.2f} °C")
print(f"T mean:    {t2m_c.mean().item():.2f} °C")
print(f"T maximum: {t2m_c.max().item():.2f} °C")

print(f"Wind maximum: {wind10.max().item():.2f} m/s")
```

```python
print("T 2%: ", t2m_c.quantile(0.02).item())
print("T 98%:", t2m_c.quantile(0.98).item())
```

```python
import xarray as xr

ds2 = xr.open_dataset("data_1.nc")

print(ds2)

for var in ds2.data_vars:
    print(f"\n--- {var} ---")
    print(ds2[var].attrs)
```

```python
ds2 = ds2.assign_coords(
    longitude=((ds2.longitude + 180) % 360) - 180
).sortby("longitude")
```

```python
print(ds2.longitude.min().item())
print(ds2.longitude.max().item())
```

```python
print(
    "Latitude match:",
    np.allclose(ds.latitude, ds2.latitude)
)

print(
    "Longitude match:",
    np.allclose(ds.longitude, ds2.longitude)
)

print(
    "Time match:",
    np.array_equal(ds.valid_time, ds2.valid_time)
)
```

```python
weather = xr.merge([ds, ds2])

weather
```

```python
print(
    weather["snowc"].min().item(),
    weather["snowc"].max().item()
)
```

```python
print("snowc:", weather.snowc.min().item(),
                  weather.snowc.max().item())

print("sde:", weather.sde.min().item(),
                weather.sde.max().item())
```
